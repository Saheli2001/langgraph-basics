```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
class BatsmanState(TypedDict):

    runs: int
    balls: int
    fours: int
    six: int

    sr: float
    bpb: float
    boundary_percenr: float
    summary: str
```

```python
graph = StateGraph(BatsmanState)
```

```python
def sr(state: BatsmanState) -> BatsmanState:

    sr = (state['runs']/ state['balls'])*100

    return {'sr':sr}
```

```python
def bpb(state: BatsmanState) -> BatsmanState:

    bpb = state['balls']/(state['fours'] + state['six'])

    return {'bpb':bpb}
```

```python
def boundary_percent(state: BatsmanState):

    boundary_percent=(state['fours']*4 + state['six']*6)/(state['runs']*100)

    return {'boundary_percent': boundary_percent}
```

```python
def summary(state: BatsmanState):

    summary = "YOU KNOWWWWW"



    return {'summary':summary}
```

```python
graph.add_node('sr',sr)
graph.add_node('bpb',bpb)
graph.add_node('boundary_percent',boundary_percent)
graph.add_node('summary',summary)


graph.add_edge(START, 'sr')
graph.add_edge(START, 'bpb')
graph.add_edge(START, 'boundary_percent')

graph.add_edge('sr', 'summary')
graph.add_edge('bpb', 'summary')
graph.add_edge('boundary_percent', 'summary')

workflow = graph.compile()

initial_state = {'runs':100,
            'balls':50,
            'fours': 6,
            'six':4}

workflow.invoke(initial_state)
```
